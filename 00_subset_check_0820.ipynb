````python
from pathlib import Path
import pandas as pd

# ------------------------------------------------------------------------------
# Configs
# ------------------------------------------------------------------------------
ROOT = Path("/content/drive/MyDrive/DAHS/datasets/3.1_subset")

HOSP_DIR = ROOT / "hosp"
ICU_DIR = ROOT / "icu"

print("File exist?")
print(ROOT.exists())
print(HOSP_DIR.exists())
print(ICU_DIR.exists())
print("==============================================")


# ------------------------------------------------------------------------------
# Load `icustays.csv.gz`
# ------------------------------------------------------------------------------
icustays_path = ICU_DIR / "icustays.csv.gz"
icustays = pd.read_csv(icustays_path)

print("`icustays.csv.gz` shape: ", icustays.shape)
print("columns: ", icustays.columns.tolist())
print("==============================================")
print(icustays.head())
print("==============================================")
print("Unique subject_id(환자 수):", icustays["subject_id"].nunique())
print("Unique hadm_id(병원 입원 episode):", icustays["hadm_id"].nunique())
print("Unique stay_id(그 안의 ICU stay):", icustays["stay_id"].nunique())
````

````python
# ------------------------------------------------------------------------------
# Load `patients/admissions.csv.gz` & check item matches well
# ------------------------------------------------------------------------------
patients_path = HOSP_DIR / "patients.csv.gz"
admissions_path = HOSP_DIR / "admissions.csv.gz"
patients = pd.read_csv(patients_path)
admissions = pd.read_csv(admissions_path)

print("==============================================")
print("patients shape   :", patients.shape)
print("admissions shape :", admissions.shape)
print("==============================================")
print("patients columns:")
print(patients.columns.tolist())
print("\nadmissions columns:")
print(admissions.columns.tolist())
print("==============================================")
print("icustays에 있는 subject_id가 patients에 모두 존재하는지?")
missing_subject = (set(icustays["subject_id"]) - set(patients["subject_id"]))
missing_hadm = (set(icustays["hadm_id"]) - set(admissions["hadm_id"]))
print("missing_subject:", len(missing_subject))
print("missing_hadm  :", len(missing_hadm))
print("==============================================")
````

````python
# ------------------------------------------------------------------------------
# Load `chartevents.csv.gz` & check event matches well
# ------------------------------------------------------------------------------
chartevents_path = ICU_DIR / "chartevents.csv.gz"
chartevents = pd.read_csv(chartevents_path)

print("==============================================")
print("chartevents shape:", chartevents.shape)
print("columns:")
print(chartevents.columns.tolist())
print("==============================================")
print(chartevents.head())
````

````python
icu_stay_ids = set(icustays["stay_id"].dropna().unique())
chart_stay_ids = set(chartevents["stay_id"].dropna().unique())

missing_in_icustays = chart_stay_ids - icu_stay_ids

print("==============================================")
print("Unique stay_id in icustays    :", len(icu_stay_ids))
print("Unique stay_id in chartevents :", len(chart_stay_ids))
print("실제 ICU 안에서 측정된 이벤트 데이터도 이 3000 stay에 맞게 잘 잘려 있나?")
print("chartevents stay_id not in icustays:", len(missing_in_icustays))
print("==============================================")
````

````python
# ------------------------------------------------------------------------------
# Load `inputevents.csv.gz` & check event matches well
# ------------------------------------------------------------------------------
inputevents_path = ICU_DIR / "inputevents.csv.gz"
inputevents = pd.read_csv(inputevents_path)

print("==============================================")
print("inputevetns shape:", inputevents.shape)
print("columns:")
print(inputevents.columns.tolist())
print("==============================================")
print(inputevents.head())
print()
print("==============================================")


icu_stay_ids = set(icustays["stay_id"].dropna().unique())
input_stay_ids = set(inputevents["stay_id"].dropna().unique())
missing_input_stays = input_stay_ids - icu_stay_ids
stays_without_inputevents = icu_stay_ids - input_stay_ids

print("Unique stay_id in icustays    :", len(icu_stay_ids))
print("Unique stay_id in inputevents :", len(input_stay_ids))
print("inputevents stay_id들이 3000 stay에 맞게 잘 잘려 있나?")
print("inputevents stay_id not in icustays:", len(missing_input_stays))
print("ICU stays without inputevents       :", len(stays_without_inputevents))
print("==============================================")
````

````python
# ------------------------------------------------------------------------------
# Load `labevents.csv.gz` & check item matches well
# ------------------------------------------------------------------------------
labevents_path = HOSP_DIR / "labevents.csv.gz"
labevents = pd.read_csv(labevents_path)

print("==============================================")
print("labevents shape:", labevents.shape)
print("columns:")
print(labevents.columns.tolist())
print("==============================================")
print(labevents.head())
````

````python
lab_subject_ids = set(labevents["subject_id"].dropna().unique())
lab_hadm_ids = set(labevents["hadm_id"].dropna().unique())

subset_subject_ids = set(patients["subject_id"].dropna().unique())
subset_hadm_ids = set(admissions["hadm_id"].dropna().unique())

lab_subject_not_in_subset = lab_subject_ids - subset_subject_ids
lab_hadm_not_in_subset = lab_hadm_ids - subset_hadm_ids

print("==============================================")
print("labevents.csv.gz의 subjects와 admissions들이 patients/admissions.csv.gz와 매칭되는가?")
print("labevents subject_id not in patients:", len(lab_subject_not_in_subset))
print("labevents hadm_id not in admissions:", len(lab_hadm_not_in_subset))
print("==============================================")
````

````python
# ------------------------------------------------------------------------------
# Load `d_items/d_labitems.csv.gz` & check item matches well
# ------------------------------------------------------------------------------
d_items_path = ICU_DIR / "d_items.csv.gz"
d_labitems_path = HOSP_DIR / "d_labitems.csv.gz"
d_items = pd.read_csv(d_items_path)
d_labitems = pd.read_csv(d_labitems_path)

print("==============================================")
print("d_items shape: ", d_items.shape)
print("d_labitems shape: ", d_labitems.shape)
print("==============================================")
print("d_items columns:")
print(d_items.columns.tolist())

print("\nd_labitems columns:")
print(d_labitems.columns.tolist())
print("==============================================")
````

````python
# ICU dictionary itemid
d_item_ids = set(d_items["itemid"].dropna().unique())

# LAB dictionary itemid
d_labitem_ids = set(d_labitems["itemid"].dropna().unique())


# chartevents에 있는데 d_items에는 없는 itemid
chart_item_ids = set(chartevents["itemid"].dropna().unique())
missing_chart_items = chart_item_ids - d_item_ids
print("Unique itemid not in chartevents: ", len(missing_chart_items))

# inputevents에 있는데 d_items에는 없는 itemid
input_item_ids = set(inputevents["itemid"].dropna().unique())
missing_input_items = input_item_ids - d_item_ids
print("Unique itemid not in d_items: ", len(missing_input_items))

# labevents에 있는데 d_labitems에는 없는 itemid
lab_item_ids = set(labevents["itemid"].dropna().unique())
missing_lab_items = lab_item_ids - d_labitem_ids
print("Unique itemid not in d_labitems: ", len(missing_lab_items))
````

````python
# ------------------------------------------------------------------------------
# Check ICU's timestamp
# ------------------------------------------------------------------------------
print("==============================================")
icustays["intime"] = pd.to_datetime(icustays["intime"])
icustays["outtime"] = pd.to_datetime(icustays["outtime"])

invalid_stay_time = icustays[icustays["outtime"] <= icustays["intime"]]
print("outtime이 intime보다 앞에 있는 경우: ", len(invalid_stay_time))
print("==============================================")
calculated_los = (icustays["outtime"] - icustays["intime"]).dt.total_seconds() / (24 * 60 * 60)

los_diff = (calculated_los - icustays["los"]).abs()
print("Max los difference: ", los_diff.max())
print("Min los difference: ", los_diff.min())
print("==============================================")
````

````python
# ------------------------------------------------------------------------------
# Check chartevent's charttime
# ------------------------------------------------------------------------------
chartevents["charttime"] = pd.to_datetime(chartevents["charttime"], errors="coerce")

print("Invalid chartevents charttime:", chartevents["charttime"].isna().sum())
````

````python
# ------------------------------------------------------------------------------
# Check inputevents's timestamp
# ------------------------------------------------------------------------------
inputevents["starttime"] = pd.to_datetime(inputevents["starttime"], errors="coerce")
inputevents["endtime"] = pd.to_datetime(inputevents["endtime"], errors="coerce")

print("Invalid input starttime:", inputevents["starttime"].isna().sum())
print("Invalid input endtime:", inputevents["endtime"].isna().sum())
print("endtime < starttime:", (inputevents["endtime"] < inputevents["starttime"]).sum())
````

````python
# inputevents에서 endtime이 starttime보다 앞에 있는 경우 조사
invalid_input_time = inputevents[inputevents["endtime"] < inputevents["starttime"]].copy()

print("Invalid rows: ", len(invalid_input_time))
print()
display(
    invalid_input_time[
        [
            "subject_id",
            "hadm_id",
            "stay_id",
            "itemid",
            "starttime",
            "endtime",
            "storetime",
            "amount",
            "amountuom",
            "rate",
            "rateuom",
            "ordercategoryname",
            "statusdescription",
        ]
    ]
)
````

````python
# ------------------------------------------------------------------------------
# Check variables for CF state annotation
# ------------------------------------------------------------------------------

# MAP 관련 variable 찾기
map_candidates = d_items[
    d_items["label"].str.contains("blood pressure mean|mean blood pressure|arterial blood pressure mean",
    case=False, na=False, regex=True)][["itemid", "label", "abbreviation", "linksto", "category", "unitname"]]

display(map_candidates)
````

````python
map_itemids = set(map_candidates["itemid"])

map_event_count = chartevents["itemid"].isin(map_itemids).sum()
map_stay_count = chartevents.loc[chartevents["itemid"].isin(map_itemids), "stay_id"].nunique()

print("MAP candidate event rows :", map_event_count)
print("MAP candidate stays      :", map_stay_count)
````

````python
# 혈압 관련 항목 전체
bp_candidates = d_items[
    d_items["label"].str.contains(
        "blood pressure",
        case=False,
        na=False
    )
][
    ["itemid", "label", "abbreviation", "linksto", "category", "unitname"]
]

display(bp_candidates)
````

````python
# Lactate 후보
lactate_candidates = d_labitems[
    d_labitems["label"].str.contains("lactate",
    case=False, na=False)][["itemid", "label", "fluid", "category"]]

display(lactate_candidates)
````

````python
lactate_itemids = set(lactate_candidates["itemid"])

lactate_event_count = labevents["itemid"].isin(lactate_itemids).sum()
lactate_subject_count = labevents.loc[
    labevents["itemid"].isin(lactate_itemids),
    "subject_id"
].nunique()
lactate_hadm_count = labevents.loc[
    labevents["itemid"].isin(lactate_itemids),
    "hadm_id"
].nunique()

print("Lactate candidate event rows :", lactate_event_count)
print("Lactate candidate subjects   :", lactate_subject_count)
print("Lactate candidate admissions :", lactate_hadm_count)
````

````python
# Vasoactive, Intropic drug 후보
drug_pattern = ("norepinephrine|epinephrine|vasopressin|"
                "dobutamine|dopamine|milrinone|phenylephrine")

drug_candidates = d_items[
    d_items["label"].str.contains(
        drug_pattern,
        case=False, na=False, regex=True
    )][["itemid", "label", "abbreviation", "linksto", "category", "unitname"]]

display(drug_candidates)
````

````python
drug_itemids = set(drug_candidates["itemid"])

drug_event_count = inputevents["itemid"].isin(drug_itemids).sum()
drug_stay_count = inputevents.loc[
    inputevents["itemid"].isin(drug_itemids),
    "stay_id"
].nunique()

print("Drug candidate event rows :", drug_event_count)
print("Drug candidate stays      :", drug_stay_count)
````
